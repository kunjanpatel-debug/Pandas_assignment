```python
import pandas as pd
```

```python
#task1

jan = pd.read_csv('swiggy_jan.csv')
feb = pd.read_csv('swiggy_feb.csv')

all_orders = pd.concat([jan, feb], ignore_index=True)
print(all_orders)
```

```python
#task2

products = pd.read_csv('fk_items.csv')
ratings = pd.read_csv('fk_ratings.csv')

merged_products = pd.concat([products, ratings], axis=1)
print(merged_products)
```

```python
#task3

spotify = pd.read_csv('spotify_daily_streams.csv')

pivoted_spotify = spotify.pivot(index='user', columns='day', values='streams')
print(pivoted_spotify)
```

```python
#task4

zomato = pd.read_csv('zomato_monthly_orders.csv')

avg_orders = zomato.pivot_table(index='restaurant', columns='month', values='total_amount', aggfunc='mean')
print(avg_orders)
```

```python
#task5


jan_sales = pd.read_csv('sales_jan.csv')
feb_sales = pd.read_csv('sales_feb.csv')
mar_sales = pd.read_csv('sales_mar.csv')

total_df = pd.concat([jan_sales, feb_sales, mar_sales], ignore_index=True)

total_sales_per_product = total_df.groupby('product')['sales'].sum().reset_index()
print(total_sales_per_product)
```
