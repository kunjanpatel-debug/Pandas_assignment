```python
import  pandas as pd
```

```python
#task1

movies = pd.DataFrame({
    'title': ['3 Idiots', 'Dangal', 'Sholay', 'Lagaan', 'Gangs of Wasseypur', 'Taare Zameen Par'],
    'rating': [8.4, 8.3, 8.1, 8.1, 8.2, 8.3],
    'release_year': [2009, 2016, 1975, 2001, 2012, 2007]
})

sorted_movies = movies.sort_values(by='rating', ascending=False)
print(sorted_movies)
```

```python
#tsk2

ipl = pd.read_csv('ipl_standings.csv')

sorted_ipl = ipl.sort_index(ascending=False)
print(sorted_ipl)
```

```python
#tqsk3

yt = pd.read_csv('youtube_trending.csv', index_col=0)

yt_reset = yt.reset_index(drop=True)
print(yt_reset)
```

```python
#task5

flipkart = pd.read_csv('flipkart_discounts.csv')

sorted_flipkart = flipkart.sort_values(by='discount', ascending=False).reset_index(drop=True)
print(sorted_flipkart)
```
