```python
import pandas as pd
```

```python
#task1


df_sales = pd.read_csv('flipkart_sales.csv')

df_sales['TotalValue'] = df_sales['Price'] * df_sales['Qty']

print(df_sales)
```

```python
#task2

df_food = pd.read_csv('food_items.csv')

df_food['DiscountedPrice'] = df_food['Price'].apply(lambda x: x * 0.90)

print(df_food)
```

```python
#task3

df_insta = pd.read_csv('instagram_posts.csv')

def format_number(val):
    if val >= 1000000:
        return f"{val / 1000000:.1f}M"
    elif val >= 1000:
        return f"{val / 1000:.1f}K"
    return str(val)

df_insta[['Likes', 'Comments']] = df_insta[['Likes', 'Comments']].map(format_number)

print(df_insta)
```

```python
#task4

df_zomato = pd.read_csv('zomato_orders_s7.csv')

df_zomato['FinalAmount'] = df_zomato.apply(lambda row: (row['Price'] * row['Qty']) + row['DeliveryCharge'], axis=1)

print(df_zomato)
```

```python
#task5


df_myntra = pd.read_csv('myntra_orders.csv')

df_myntra['GST'] = (df_myntra['Price'] * df_myntra['Qty']) * 0.05

print(df_myntra)
```
