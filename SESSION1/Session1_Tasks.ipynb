```python
#Task3

followers_list = [1200, 54000, 780, 150000, 32000]

followers = pd.Series(followers_list)

max_idx = followers.idxmax()
min_idx = followers.idxmin()

print(f"Highest follower count: {followers[max_idx]} (at index {max_idx})")
print(f"Lowest follower count: {followers[min_idx]} (at index {min_idx})")
```
