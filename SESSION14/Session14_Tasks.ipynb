```python
#task2

ipl_df = pd.read_csv('ipl_matches_s14.csv')

ipl_clean = ipl_df.dropna(subset=['winner']).reset_index(drop=True)

print("Rows before cleaning:", len(ipl_df))
print("Rows after cleaning:", len(ipl_clean))
print(ipl_clean.head())
```

```python
#task3

ipl_clean = pd.read_csv('ipl_matches_s14.csv').dropna(subset=['winner']).reset_index(drop=True)

def get_margin_type(row):
    if row['win_by_runs'] > 0:
        return 'Runs'
    elif row['win_by_wickets'] > 0:
        return 'Wickets'
    return 'Tie/No Result'

ipl_clean['match_margin_type'] = ipl_clean.apply(get_margin_type, axis=1)

print(ipl_clean['match_margin_type'].value_counts())
print(ipl_clean[['winner', 'win_by_runs', 'win_by_wickets', 'match_margin_type']].head())
```

```python
#task4


ipl_clean = pd.read_csv('ipl_matches_s14.csv').dropna(subset=['winner']).reset_index(drop=True)
teams_df = pd.read_csv('ipl_teams_s14.csv')

merged = pd.merge(ipl_clean, teams_df, left_on='team1', right_on='team', how='left')
merged = merged.rename(columns={'home_city': 'team1_city'}).drop(columns=['team'])

merged = pd.merge(merged, teams_df, left_on='team2', right_on='team', how='left')
merged = merged.rename(columns={'home_city': 'team2_city'}).drop(columns=['team'])

print(merged[['match_id', 'team1', 'team1_city', 'team2', 'team2_city', 'winner']].head(10))
```

```python
#task5

ipl_clean = pd.read_csv('ipl_matches_s14.csv').dropna(subset=['winner']).reset_index(drop=True)

wins_pivot = ipl_clean.pivot_table(index='winner', columns='season', values='match_id', aggfunc='count', fill_value=0)

wins_pivot['Total_Wins'] = wins_pivot.sum(axis=1)

top_3_teams = wins_pivot.sort_values(by='Total_Wins', ascending=False).head(3)

print("--- Wins per Season ---")
print(wins_pivot)

print("\nTop 3 Teams Overall")
print(top_3_teams)
```
