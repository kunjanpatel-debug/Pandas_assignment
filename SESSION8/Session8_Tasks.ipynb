```python
import pandas as pd
```

```python
#task1


df_insta = pd.read_csv('instagram.csv')

df_insta['username'] = df_insta['username'].str.lower()

print(df_insta)
```

```python
#task2


df_spotify = pd.read_csv('spotify_songs_remix.csv')

df_spotify['song_title'] = df_spotify['song_title'].str.replace('Remix', '', regex=False).str.strip()

# 3. Display cleaned series
print(df_spotify['song_title'])
```

```python
#task3

df_zomato = pd.read_csv('zomato_reviews_text.csv')

delivery_reviews = df_zomato[df_zomato['review'].str.contains('delivery', case=False, na=False)]

print(delivery_reviews[['restaurant', 'review']])
```

```python
#task4


df_flipkart = pd.read_csv('flipkart_categories.csv')

df_flipkart['MainCategory'] = df_flipkart['category_path'].str.split('/').str[0]

print(df_flipkart[['category_path', 'MainCategory']])
```

```python
#task5

df_yt = pd.read_csv('youtube_titles.csv')

df_yt['title'] = df_yt['title'].str.upper().str.replace(' ', '_')

print(df_yt)
```
