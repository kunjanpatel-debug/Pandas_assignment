```python
import pandas as pd
```

```python
#task1

date_series = pd.Series(['15-08-2023', '26-01-2024', '02-10-2023', '14-11-2023', '25-12-2023'])

datetime_series = pd.to_datetime(date_series, dayfirst=True)

print(datetime_series)
```

```python
#task2

df_movies = pd.read_csv('movie_releases.csv')

df_movies['release_date'] = pd.to_datetime(df_movies['release_date'], dayfirst=True)

df_movies['year'] = df_movies['release_date'].dt.year
df_movies['month'] = df_movies['release_date'].dt.month
df_movies['day'] = df_movies['release_date'].dt.day

print(df_movies)
```
