```python
import pandas as pd
```

```python

movies_df = pd.read_csv('movies.csv')

selected_movies = movies_df.loc[:, ['title', 'rating']]
print(selected_movies)
```

```python
#task2

flipkart_df = pd.read_csv('flipkart_products.csv')

first_10_products = flipkart_df.iloc[0:10, :]
print(first_10_products)
```

```python
#task3

zomato_df = pd.read_csv('zomato_orders.csv')

high_value_orders = zomato_df[zomato_df['order_amount'] > 500]
print(high_value_orders)
```

```python
#task4

spotify_df = pd.read_csv('spotify_songs.csv')

filtered_songs = spotify_df.query('streams > 1000000 and duration < 180')
print(filtered_songs)
```

```python
#task5

ipl_runs_df = pd.read_csv('ipl_runs.csv')

filtered_matches = ipl_runs_df[(ipl_runs_df['total_runs'] >= 180) & (ipl_runs_df['total_runs'] <= 220)]
print(filtered_matches)
```
