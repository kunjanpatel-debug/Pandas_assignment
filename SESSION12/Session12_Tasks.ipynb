```python
import pandas as pd
```

```python
#task1

df_ipl = pd.read_csv('ipl_team_runs.csv')

team_total_runs = df_ipl.groupby('team')['runs'].sum().reset_index()

print(team_total_runs)
```

```python
#task2

df_zomato = pd.read_csv('zomato_city_orders.csv')

avg_order_df = df_zomato.groupby(['city', 'restaurant'])['order_amount'].agg('mean').reset_index()

print(avg_order_df)
```

```python
#task3

df_fk = pd.read_csv('flipkart_sales_agg.csv')

multi_agg_df = df_fk.groupby(['category', 'region']).agg({
    'sales': 'sum',
    'units_sold': 'mean'
}).reset_index()

print(multi_agg_df)
```

```python
#task4

df_rev = pd.read_csv('regional_revenue.csv')

revenue_summary = df_rev.groupby('region').agg({
    'revenue': ['sum', 'max', 'min']
}).reset_index()

# 3. Display result
print(revenue_summary)
```
