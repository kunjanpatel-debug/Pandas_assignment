```python
import pandas as pd
```

```python
#task1

df_players = pd.read_csv('ipl_players_missing.csv')

missing_age_rows = df_players[df_players['age'].isna()]
print("Rows with missing age:")
print(missing_age_rows)
```

```python
#task2

df_orders = pd.read_csv('food_orders_missing.csv')

cleaned_orders = df_orders.dropna(subset=['delivery_rating'])
print("Cleaned DataFrame (after dropping missing ratings):")
print(cleaned_orders)
```

```python
#task3

df_reviews = pd.read_csv('flipkart_reviews_missing.csv')

mean_rating = df_reviews['rating'].mean()
df_reviews['rating'] = df_reviews['rating'].fillna(mean_rating)

print(f"Mean Rating Used: {mean_rating}")
print("Updated DataFrame:")
print(df_reviews)
```

```python
#task4

df_spotify = pd.read_csv('spotify_playlist_missing.csv')

median_duration = df_spotify['duration'].median()
df_spotify['duration'] = df_spotify['duration'].fillna(median_duration)

df_spotify.to_csv('cleaned_spotify_playlist.csv', index=False)

print(f"Median Duration Used: {median_duration}")
print(df_spotify)
```

```python
#task5


df_insta = pd.read_csv('instagram_users_missing.csv')

df_insta['followers'] = df_insta['followers'].fillna(0)

print("Result after filling missing followers with 0:")
print(df_insta)
```
