```python
#task2

df_ipl = pd.read_csv('IPL_2023_Batsmen.csv')

mean_runs = df_ipl['runs'].mean()
median_runs = df_ipl['runs'].median()
mode_runs = df_ipl['runs'].mode()[0]
min_runs = df_ipl['runs'].min()
max_runs = df_ipl['runs'].max()

print("Runs Univariate Analysis")
print("Mean:", mean_runs)
print("Median:", median_runs)
print("Mode:", mode_runs)
print("Minimum:", min_runs)
print("Maximum:", max_runs)
```
