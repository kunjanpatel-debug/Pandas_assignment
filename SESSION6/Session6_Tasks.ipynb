```python
import pandas as pd
```

```python
#task1

df_spotify = pd.read_csv('spotify_streams_duplicates.csv')

df_spotify_cleaned = df_spotify.drop_duplicates(subset=['song'], keep='first')

print("Original Data Count:", len(df_spotify))
print("Cleaned Data Count:", len(df_spotify_cleaned))
print(df_spotify_cleaned)
```

```python
#task2

df_reviews = pd.read_csv('flipkart_reviews_duplicates.csv')

dup_before = df_reviews.duplicated(subset=['product_id', 'user_id']).sum()
print("Number of duplicate reviews BEFORE removal:", dup_before)

df_reviews_cleaned = df_reviews.drop_duplicates(subset=['product_id', 'user_id'])

dup_after = df_reviews_cleaned.duplicated(subset=['product_id', 'user_id']).sum()
print("Number of duplicate reviews AFTER removal:", dup_after)

print("\nCleaned Reviews DataFrame:")
print(df_reviews_cleaned)
```

```python
#task3

df_zomato = pd.read_csv('zomato_outliers.csv')

q1 = df_zomato['order_amount'].quantile(0.25)
q3 = df_zomato['order_amount'].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df_zomato[(df_zomato['order_amount'] < lower_bound) | (df_zomato['order_amount'] > upper_bound)]

print(f"IQR: {iqr} | Lower Bound: {lower_bound} | Upper Bound: {upper_bound}")
print("Outlier Order IDs:")
print(outliers[['order_id', 'restaurant', 'order_amount']])
```

```python
#task4

from scipy import stats

df_tickets = pd.read_csv('ipl_ticket_sales.csv')
df_tickets['z_score'] = stats.zscore(df_tickets['tickets_sold'])

zscore_outliers = df_tickets[df_tickets['z_score'].abs() > 2]

print("Dates with extreme ticket sales (|Z-score| > 2):")
print(zscore_outliers[['date', 'tickets_sold', 'z_score']])
```

```python
#task5

from scipy import stats

def find_wallet_outliers(df, column_name='amount', threshold=2):
    z_scores = stats.zscore(df[column_name])
    return df[abs(z_scores) > threshold]

df_paytm = pd.read_csv('paytm_wallet.csv')
outlier_transactions = find_wallet_outliers(df_paytm, column_name='amount', threshold=2)

print("--- PAYTM WALLET OUTLIER TRANSACTIONS ---")
print(outlier_transactions)
```
