```python
import pandas as pd
```

```python
#task1

users = pd.read_csv('users.csv')
orders = pd.read_csv('orders.csv')

inner_result = pd.merge(users, orders, on='user_id', how='inner')
print(inner_result)
```

```python
#task2

users = pd.read_csv('users.csv')
orders = pd.read_csv('orders.csv')

left_result = pd.merge(users, orders, on='user_id', how='left')
print(left_result)
```
